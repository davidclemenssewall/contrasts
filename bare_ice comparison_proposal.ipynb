```python
# imports
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Read in a1 data
data_path = os.path.join('..', 'data', 'whoi_l_arm_transmittance')

project_names = ['20250710-r1-coring',
                 '20250715-r2-coring',
                 '20250715-r2-lead5',
                 '20250715-r2-lead20',
                 '20250720-r3-coring',
                 '20250720-r3-lead_far',
                 '20250726-r1-coring',
                 '20250726-r1-lead_close',
                 '20250726-r1-lead_far',
                 '20250730-r2-coring',
                 '20250730-r2-ssl',
                 '20250730-r2-lead50',
                 '20250804-r3-coring',
                 ]

ls_df_a1 = []
for project_name in project_names:
    file_path = os.path.join(data_path,
                            project_name,
                            'contrasts_l_arm_' + project_name + '.a1.h5')
    ls_df_a1.append(pd.read_hdf(file_path))

df_a1 = pd.concat(ls_df_a1)
```

```python
df_a1
```

```python
# Plot all data, just for fun
f, axs = plt.subplots(2, 1, sharex=True, figsize=(8,8))

sns.lineplot(x='wavelength_nm', y='value', hue='timestamp_utc', estimator=None,
             data=df_a1.query('variable=="Albedo"'), ax=axs[0], legend=False)

sns.lineplot(x='wavelength_nm', y='value', hue='timestamp_utc', estimator=None,
             data=df_a1.query('variable=="Transmittance"'), ax=axs[1], legend=False)
```

```python
# Plot all data, just for fun
f, axs = plt.subplots(2, 1, sharex=True, figsize=(6,6))

location = 'coring'
station = '1'

sns.lineplot(x='wavelength_nm', y='value', hue='timestamp_utc', estimator=None,
             data=df_a1.query('variable=="Albedo" & location==@location & station==@station'), 
             ax=axs[0], legend=False, palette='viridis')

sns.lineplot(x='wavelength_nm', y='value', hue='timestamp_utc', estimator=None,
             data=df_a1.query('variable=="Transmittance" & location==@location & station==@station'), 
             ax=axs[1], legend=False, palette='viridis')

axs[0].set_ylabel('Albedo')
axs[1].set_ylabel('Transmittance')

f.suptitle(f'{location} site, station: {station}')
```
